```python
import torch
from transformers import WhisperForConditionalGeneration, WhisperProcessor

print("Cihaz yoxlanılır...")
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"İstifadə olunan cihaz: {device}")

model_id = "openai/whisper-base"

print(f"'{model_id}' üçün processor yüklənir...")
processor = WhisperProcessor.from_pretrained(model_id, language="azerbaijani", task="transcribe")

print(f"'{model_id}' üçün model çəkiləri yüklənir (bu bir az vaxt ala bilər)...")
model = WhisperForConditionalGeneration.from_pretrained(model_id).to(device)

print("Yaddaş optimizasiyası aktiv edilir...")
model.config.use_cache = False
model.gradient_checkpointing_enable()

print("HAZIRDIR!")
```
